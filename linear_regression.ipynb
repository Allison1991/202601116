```python
# 1. 載入所需套件
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 2. 準備極簡資料
# 特徵 X (輸入值)，並重塑為 scikit-learn 要求的形狀 (-1, 1)
X = np.array([1, 2, 3, 4, 5, 6, 7]).reshape(-1, 1) 
# 目標 y (輸出值)
y = np.array([3.2, 4.8, 7.1, 8.5, 11.2, 12.5, 15.1]) 

# 3. 訓練模型
model = LinearRegression()
model.fit(X, y)

# 4. 獲取參數並進行預測
slope = model.coef_[0]        # 斜率
intercept = model.intercept_  # 截距
y_pred = model.predict(X)     # 使用模型預測 X 值的結果

# 5. 輸出文字結果
print("--- 線性迴歸模型結果 ---")
print(f"原始數據點數量: {len(X)}")
print(f"迴歸方程: Y = {slope:.3f} * X + {intercept:.3f}")
print("-" * 25)

# 6. 視覺化結果
plt.figure(figsize=(8, 5))

# 繪製原始數據點
plt.scatter(X, y, color='blue', s=100, label='原始數據點')

# 繪製迴歸直線
plt.plot(X, y_pred, color='red', linewidth=3, linestyle='--', 
         label=f'迴歸直線 (Y = {slope:.3f}X + {intercept:.3f})')

# 設定圖表
plt.title('線性迴歸模型擬合結果', fontsize=16)
plt.xlabel('X 軸 (輸入特徵)', fontsize=12)
plt.ylabel('Y 軸 (目標值)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.6)
plt.show()
```
